```python
# Task 4: Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is performed on the prepared Supermarket Sales dataset to understand the data, identify important characteristics, discover patterns and trends, and find relationships between variables. The analysis includes descriptive statistics, data inspection, data cleaning, missing-value handling, correlation analysis, identification of important patterns and trends, and detection of outliers where applicable. The findings from the EDA will help in understanding sales performance, customer behavior, product-line performance, and other important aspects of the supermarket transactions.
```

```python
##  Descriptive Statistics
# Display descriptive statistics for all numerical columns
display(data.describe())
```

```python
##  Data Inspection
# Display the first 5 records to inspect the dataset after preparation
display(data.head())

# Display the last 5 records to inspect the end of the dataset
display(data.tail())

# Display the number of rows and columns in the dataset
print("Dataset Shape:", data.shape)
```

```python
##  Data Cleaning
# Check the dataset for duplicate records
duplicate_count = data.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

# Check for missing values in all columns
print("\nMissing values in each column:")
print(data.isnull().sum())
```

```python
##  Missing-Value Handling
# Count missing values in each column
missing_values = data.isnull().sum()

print("Missing values before handling:")
print(missing_values)

# Calculate the total number of missing values
total_missing = missing_values.sum()

print("\nTotal missing values:", total_missing)
```

```python
##  Correlation Analysis
# Select numerical columns for correlation analysis
numeric_data = data.select_dtypes(include="number")

# Calculate the correlation between numerical variables
correlation_matrix = numeric_data.corr()

# Display the correlation matrix
display(correlation_matrix)
```

```python
##  Important Patterns and Trends
# Calculate total sales for each product line
product_sales = data.groupby("Product line")["Total"].sum().sort_values(ascending=False)

print("Total Sales by Product Line:")
display(product_sales)

# Calculate total sales for each city
city_sales = data.groupby("City")["Total"].sum().sort_values(ascending=False)

print("\nTotal Sales by City:")
display(city_sales)

# Calculate total sales for each month
monthly_sales = data.groupby(data["Date"].dt.month)["Total"].sum()

print("\nTotal Sales by Month:")
display(monthly_sales)
```

```python
##  Outlier Detection
# Check for outliers in important numerical sales variables using the IQR method
numeric_columns = ["Unit price", "Quantity", "Total", "Gross income"]

for column in numeric_columns:
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[(data[column] < lower_limit) | (data[column] > upper_limit)]

    print(f"{column}: {len(outliers)} outliers")
```

```python
##  EDA Findings

The Exploratory Data Analysis provided an overall understanding of the Supermarket Sales dataset. Descriptive statistics helped summarize the numerical variables, while data inspection confirmed the structure and quality of the prepared dataset. Data cleaning and missing-value checks were performed to ensure the dataset was suitable for analysis. Correlation analysis was used to identify relationships between numerical variables. Product-line, city, and monthly sales analysis helped identify important sales patterns and trends. Outlier detection using the IQR method was performed for selected numerical variables to identify unusually high or low observations.

The specific findings from these analyses will be used to support the visualizations, Power BI dashboard, business insights, and recommendations in the following tasks.
```

```python
# Task 5: Data Visualization

Data visualizations are created using the prepared Supermarket Sales dataset to present important information in a clear and understandable way.
The visualizations help compare sales performance across product lines and cities, identify monthly sales trends, understand payment method usage, and examine relationships between numerical variables.
Each visualization includes a meaningful title and appropriate labels.
```

```python
##  Total Sales by Product Line
# Import matplotlib for creating visualizations
import matplotlib.pyplot as plt

# Calculate total sales for each product line
product_line_sales = data.groupby("Product line")["Total"].sum().sort_values(ascending=False)

# Create a bar chart to compare sales across product lines
plt.figure(figsize=(10, 6))
product_line_sales.plot(kind="bar")

# Add a meaningful title and axis labels
plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")

# Rotate product-line names for better readability
plt.xticks(rotation=45, ha="right")

# Adjust the layout to prevent labels from being cut off
plt.tight_layout()

# Display the chart
plt.show()
```

```python
##  Total Sales by City
# Calculate total sales for each city
city_sales = data.groupby("City")["Total"].sum().sort_values(ascending=False)

# Create a bar chart to compare sales across cities
plt.figure(figsize=(8, 5))
city_sales.plot(kind="bar")

# Add a meaningful title and axis labels
plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")

# Adjust the layout for better presentation
plt.tight_layout()

# Display the chart
plt.show()
```

```python
##  Monthly Sales Trend
# Calculate total sales for each month
monthly_sales = data.groupby(data["Date"].dt.month)["Total"].sum()

# Create a line chart to show the monthly sales trend
plt.figure(figsize=(8, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

# Add a meaningful title and axis labels
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

# Display month numbers on the x-axis
plt.xticks([1, 2, 3], ["January", "February", "March"])

# Add a grid to make the trend easier to understand
plt.grid(True)

# Adjust the layout for better presentation
plt.tight_layout()

# Display the chart
plt.show()
```

```python
##  Payment Method Distribution
# Count the number of transactions for each payment method
payment_counts = data["Payment"].value_counts()

# Create a pie chart to show the distribution of payment methods
plt.figure(figsize=(7, 7))
plt.pie(
    payment_counts.values,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

# Add a meaningful title to the chart
plt.title("Payment Method Distribution")

# Display the chart
plt.show()
```

```python
##  Unit Price vs Quantity
# Create a scatter plot to examine the relationship between unit price and quantity
plt.figure(figsize=(8, 5))
plt.scatter(data["Unit price"], data["Quantity"], alpha=0.6)

# Add a meaningful title and axis labels
plt.title("Unit Price vs Quantity")
plt.xlabel("Unit Price")
plt.ylabel("Quantity")

# Adjust the layout for better presentation
plt.tight_layout()

# Display the chart
plt.show()
```
